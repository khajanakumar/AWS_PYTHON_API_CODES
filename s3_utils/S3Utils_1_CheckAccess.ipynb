```python
import boto3
from botocore.exceptions import NoCredentialsError,ClientError,BotoCoreError
```

```python
class S3UtilsCheckAccess:

   @staticmethod
   def test_s3_access(bucket_name, access_key, secret_key):
    """
    Test whether the provided AWS credentials can access an S3 bucket.

    This function creates an S3 client using the supplied AWS access key
    and secret key, then uses the S3 HeadBucket operation to verify that
    the specified bucket can be accessed.

    Args:
        bucket_name (str):
            Name of the S3 bucket to test access against.

        access_key (str):
            AWS access key ID used to authenticate with Amazon S3.

        secret_key (str):
            AWS secret access key associated with the access key ID.

    Returns:
        tuple[bool, str]:
            A tuple containing:

            - bool:
                True if the bucket can be accessed successfully,
                otherwise False.

            - str:
                A descriptive success or error message.

    Notes:
        The AWS region is currently set to "us-east-1".

        The function handles common AWS authentication, client, and SDK
        errors and returns them as part of the result rather than raising
        the exceptions to the caller.

    Example:
        success, message = test_s3_access(
            bucket_name="my-bucket",
            access_key="AKIA...",
            secret_key="..."
        )

        if success:
            print(message)
        else:
            print(f"S3 access failed: {message}")
    """

    try:
        s3 = boto3.client(
            "s3",
            aws_access_key_id=access_key,
            aws_secret_access_key=secret_key,
            region_name=None
        )

        # Test access to the bucket
        s3.head_bucket(Bucket=bucket_name)

        return True, f"Access successful to bucket '{bucket_name}'"

    except NoCredentialsError:
        return False, "AWS credentials are invalid or could not be found."

    except ClientError as e:
        error = e.response.get("Error", {})
        code = error.get("Code", "Unknown")
        message = error.get("Message", str(e))

        return False, f"AWS error {code}: {message}"

    except BotoCoreError as e:
        return False, f"AWS SDK error: {str(e)}"

    except Exception as e:
        return False, f"Unexpected error: {str(e)}"

    
    
     
```
