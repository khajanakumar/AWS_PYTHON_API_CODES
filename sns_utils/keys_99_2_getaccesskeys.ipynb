```python
# fetchone example

import sqlite3

sqlite_conn = sqlite3.connect('keys.db')
sqlite_cursor = sqlite_conn.cursor()

# Using fetchone() to get results one by one
sqlite_cursor.execute('''SELECT bucketname, accesskey, secretkey FROM keys;''')
print("Key Values  database:")
row = sqlite_cursor.fetchone()  # Fetch one row at a time
print(row)
bucketname,keyname,secretkey = row

print("bucketname->", bucketname)
print("keyname->",keyname)
print("secretkey->",secretkey)

sqlite_cursor.close()
sqlite_conn.commit()
sqlite_conn.close()
```
